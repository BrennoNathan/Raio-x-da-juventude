```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Raio-X da Juventude iniciado!")
```

```python
dados = {
    "jovem": list(range(1, 16)),
    "acesso_internet": [
        "Sim", "Sim", "Não", "Sim", "Não",
        "Sim", "Sim", "Não", "Sim", "Sim",
        "Não", "Sim", "Sim", "Não", "Sim"
    ],
    "horas_estudo": [
        2, 3, 1, 2, 1,
        4, 3, 2, 2, 5,
        1, 3, 4, 2, 3
    ],
    "frequencia_escolar": [
        "Sim", "Sim", "Sim", "Sim", "Não",
        "Sim", "Sim", "Não", "Sim", "Sim",
        "Sim", "Sim", "Sim", "Não", "Sim"
    ]
}

df = pd.DataFrame(dados)

df
```

```python
total_jovens = len(df)

com_internet = (df["acesso_internet"] == "Sim").sum()

sem_internet = (df["acesso_internet"] == "Não").sum()

percentual_internet = com_internet / total_jovens * 100

media_estudo = df["horas_estudo"].mean()

print("===== RAIO-X DA JUVENTUDE =====")
print("Total de jovens:", total_jovens)
print("Com internet:", com_internet)
print("Sem internet:", sem_internet)
print(f"Percentual com internet: {percentual_internet:.1f}%")
print(f"Média de horas de estudo por dia: {media_estudo:.1f}")
```

```python
contagem = df["acesso_internet"].value_counts()

plt.figure(figsize=(7, 4))

sns.barplot(
    x=contagem.index,
    y=contagem.values,
    color="steelblue"
)

plt.title("Acesso à internet entre os jovens")
plt.xlabel("Possui acesso à internet?")
plt.ylabel("Quantidade de jovens")

plt.show()
```

```python
frequencia = df["frequencia_escolar"].value_counts()

print("Frequência escolar:")
print(frequencia)

plt.figure(figsize=(7, 4))

sns.countplot(
    data=df,
    x="frequencia_escolar",
    color="cornflowerblue"
)

plt.title("Frequência escolar dos jovens")
plt.xlabel("Frequenta a escola?")
plt.ylabel("Quantidade de jovens")

plt.show()
```
